```python
import numpy as np
import matplotlib.pyplot as plt

# Data
x_all = np.linspace(0, 3, 600)
x_pos = np.linspace(0.05, 3, 600)  # for ln(x)

y_ln = np.log(x_pos)
y_exp = np.exp(x_all)
y_x4 = x_all**4

# Plot
fig, ax = plt.subplots(figsize=(7.8, 5.2), dpi=150)

ax.plot(x_pos, y_ln, label="R", linewidth=2)                 # red (default) -> we'll set colors explicitly below
ax.plot(x_all, y_exp, label="G", linewidth=2, linestyle="-.")
ax.plot(x_all, y_x4, label="B", linewidth=2, linestyle="--")

# Colors to match the figure
ax.lines[0].set_color("red")
ax.lines[1].set_color("limegreen")
ax.lines[2].set_color("blue")

# Axes, ticks, grid
ax.set_xlim(0, 3)
ax.set_ylim(-2, 10)
ax.set_xticks(np.arange(0, 3.01, 0.5))
ax.set_yticks(np.arange(-2, 10.01, 2))
ax.grid(True, linewidth=1, alpha=0.6)

# Legend (boxed, upper-right)
ax.legend(loc="upper right", frameon=True, framealpha=1, fancybox=False, edgecolor="black", fontsize=14)

# Make axes look like arrows (approximate the style)
for spine in ["top", "right"]:
    ax.spines[spine].set_visible(False)

ax.annotate("", xy=(3, -2), xytext=(0, -2),
            arrowprops=dict(arrowstyle="->", linewidth=1.2, color="black"))
ax.annotate("", xy=(0, 10), xytext=(0, -2),
            arrowprops=dict(arrowstyle="->", linewidth=1.2, color="black"))

# Hide bottom/left spines since we drew arrow axes
ax.spines["bottom"].set_visible(False)
ax.spines["left"].set_visible(False)

plt.savefig("../2526-2/assignments/img/01/functions_plot.svg", bbox_inches="tight")
plt.show()
```
